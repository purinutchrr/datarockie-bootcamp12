```python
# TODO 06 - which Segment has the highest profit in 2018
df_2018 = df[df['Order Date'].dt.year == 2018]

df_2018.groupby(['Segment'])[['Profit']].sum()
```

```python
# TODO 07 - which top 5 States have the least total sales between 15 April 2019 - 31 December 2019
date = df[(df['Order Date'] >= '2019-04-15') & (df['Order Date'] <= '2019-12-31')]
```

```python
date[['State', 'Sales']].sort_values(['Sales'], ascending = True).head()
```

```python
# TODO 08 - what is the proportion of total sales (%) in West + Central in 2019 e.g. 25% 
df_2019 = df[df['Order Date'].dt.year == 2019]
```

```python
total_sales = df_2019['Sales'].sum()
```

```python
west_central_sales = df_2019[df_2019['Region'].isin(['West', 'Central'])]['Sales'].sum()
```

```python
proportion = west_central_sales / total_sales * 100
print(f'Proportion of total sales in West + Central in 2019: {proportion:.2f}%')
```

```python
# TODO 09 - find top 10 popular products in terms of number of orders vs. total sales during 2019-2020
df_2019_2020 = df[df['Order Date'].dt.year.isin([2019, 2020])]
```

```python
result = df_2019_2020.groupby(['Product Name']).agg({'Order ID': 'count', 'Sales': 'sum'})
result = result.rename(columns = {'Order ID': 'Number of Order'})
top_10_order = result.sort_values(['Number of Order'], ascending = False).head(10)
top_10_order
```

```python
top_10_sales = result.sort_values(['Sales'], ascending = False).head(10)
top_10_sales
```

```python
# TODO 10 - plot at least 2 plots, any plot you think interesting :)
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
fig, axes = plt.subplots(1, 2, figsize = (14, 7))
sns.barplot(data = top_10_order, x = 'Product Name', y = 'Number of Order', ax = axes[0])
axes[0].set_xticklabels(axes[0].get_xticklabels(), rotation = 45, horizontalalignment ='right')
sns.barplot(data = top_10_sales, x = 'Product Name', y = 'Sales', ax = axes[1])
axes[1].set_xticklabels(axes[1].get_xticklabels(), rotation = 45, horizontalalignment ='right')
plt.show()
```
