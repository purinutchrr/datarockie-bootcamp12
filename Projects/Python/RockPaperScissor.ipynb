# PYC Game

```python
import random
```

```python
def pyc_game(user, computer):
    if user == computer:
        return 'Draw'
    elif (user == 'rock' and computer == 'scissors') or \
        (user == 'scissors' and computer == 'paper') or \
        (user == 'paper' and computer == 'rock'):
        return 'You win'
    else:
        return 'You lose'

user_point = 0
computer_point = 0

while True:
    user = input('My choice(rock, scissors, paper, stop): ')
    if user == 'stop':
        break

    if user not in ['rock', 'scissors', 'paper']:
        print('Invalid choice')
        continue

    computer = random.choice(['rock', 'scissors', 'paper'])
    print(f'Computer choice: {computer}')

    result = pyc_game(user, computer)
    print(result)

    if result == 'You win':
        user_point += 1
    elif result == 'You lose':
        computer_point += 1
    else:
        pass

print('-' * 20)
print(f'My point: {user_point}')
print(f'Computer point: {computer_point}')
```
