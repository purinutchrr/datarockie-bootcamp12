```python
pizza_menu = {
    'menu': ['hawaiian', 'seafood', 'parma', 'truffle', 'doublecheese'],
    'price': [200, 350, 250, 400, 300]
}

pizza_size = {
    'small': 1,
    'medium': 1.2,
    'large': 1.5
}

total_bill = 0
orders_list = []

def get_pizza_price(name, size):
    position = pizza_menu['menu'].index(name)
    base_price = pizza_menu['price'][position]
    multiplier = pizza_size[size]
    result = base_price * multiplier
    return result

print('Welcome to my pizza shop!')
customer_name = input('What is your name: ')

while True:
    print(f"\n--- Current Menu: {pizza_menu['menu']} ---")
    cust_choice = input('Select pizza menu (or type "done" to checkout): ').lower()

    if cust_choice == 'done':
        break

    if cust_choice in pizza_menu['menu']:
        size_choice = input('Select size (small, medium, large): ').lower()

        if size_choice in pizza_size:
            price = get_pizza_price(cust_choice, size_choice)

            total_bill += price
            orders_list.append(f"{cust_choice} ({size_choice})")

            print(f"Added {cust_choice} to your cart. Current total: {total_bill} Baht")
        else:
            print("Invalid size!")
    else:
        print("Menu not found!")

print('\n' + '='*20)
print(f"Receipt for: {customer_name}")
print("Items ordered:", orders_list)
print(f"Grand Total: {total_bill} Baht")
print("Thank you for your order!")
```
